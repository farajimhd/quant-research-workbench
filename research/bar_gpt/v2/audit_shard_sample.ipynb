# BarGPT real shard sample audit

Loads one deterministic random block from a certified workstation shard and displays the exact model-ready inputs, causal context indices, autoregressive targets, physical-horizon targets, and optional checkpoint predictions. No ClickHouse writes or shard changes occur.

```python
from pathlib import Path
import importlib, os, random, sys
import polars as pl
try:
    from IPython.display import display
except ImportError:
    display = print  # Allows the same read-only audit cells to run as a plain Python smoke.

REPO_CANDIDATES = (Path(r'D:/TradingML/codes/quant-research-workbench'), Path(r'D:/TradingCodes/quant-research-workbench'))
REPO = next((path for path in REPO_CANDIDATES if path.exists()), None)
if REPO is None: raise RuntimeError(f'BarGPT repository not found: {REPO_CANDIDATES}')
REPO = REPO.resolve()
sys.dont_write_bytecode = True
repo_text = str(REPO)
sys.path[:] = [entry for entry in sys.path if entry != repo_text]
sys.path.insert(0, repo_text)
# A notebook kernel can retain pre-sync BarGPT modules indefinitely. Reload the
# complete research package only when it came from another repository; otherwise
# evict the BarGPT subtree so this run reads the synchronized source files.
loaded_research = sys.modules.get('research')
loaded_research_file = Path(getattr(loaded_research, '__file__', '') or '.').resolve() if loaded_research else None
expected_research = (REPO / 'research').resolve()
wrong_research_root = loaded_research_file is not None and not loaded_research_file.is_relative_to(expected_research)
module_prefixes = ('research.', 'research') if wrong_research_root else ('research.bar_gpt.', 'research.bar_gpt')
for module_name in tuple(sys.modules):
    if any(module_name == prefix or module_name.startswith(prefix) for prefix in module_prefixes):
        sys.modules.pop(module_name, None)
importlib.invalidate_caches()
from research.bar_gpt.v2.offline_shards import OFFLINE_SHARD_CONTRACT_VERSION, verify_shard_catalog_lock
SHARD_OVERRIDE = os.environ.get('BAR_GPT_SHARD_ROOT', '').strip()
SHARD_CANDIDATES = tuple(dict.fromkeys(path for path in (
    Path(SHARD_OVERRIDE) if SHARD_OVERRIDE else None,
    Path(r'D:/TradingML/runtimes/bar_gpt/v2/offline_shards_v12'),
    Path(r'\\DESKTOP-SAAI85T/Workstation-D/TradingML/runtimes/bar_gpt/v2/offline_shards_v12'),
) if path is not None))
verified_catalogs = {}
catalog_errors = {}
for path in SHARD_CANDIDATES:
    try:
        verified_catalogs[path] = verify_shard_catalog_lock(path)
    except RuntimeError as exc:
        catalog_errors[path] = str(exc)
SHARD_ROOT = next(iter(verified_catalogs), None)
if SHARD_ROOT is None: raise RuntimeError(
    f'No audited immutable contract-{OFFLINE_SHARD_CONTRACT_VERSION} BarGPT v12 catalog found; errors={catalog_errors}'
)
CATALOG_LOCK = verified_catalogs[SHARD_ROOT]
CHECKPOINT_TEXT = ''  # Optional checkpoint_latest.pt for prediction columns
CHECKPOINT = Path(CHECKPOINT_TEXT).expanduser() if CHECKPOINT_TEXT.strip() else None
SEED = 17
TICKERS = ('AAPL',)  # Set () for a deterministic sample across the full catalog (slower).
TAIL_CONTEXT_ROWS = 8
print('repository:', REPO)
print('loader contract:', OFFLINE_SHARD_CONTRACT_VERSION)
print('shards:', SHARD_ROOT)
print('catalog locked:', CATALOG_LOCK['locked_at'])
print('catalog units:', f"{int(CATALOG_LOCK['catalog_counts']['units']):,}")
print('catalog sha256:', CATALOG_LOCK['catalog_sha256'])
```

```python
from research.bar_gpt.v2.shard_data_audit import (
    context_rows, data_config_for_sample, load_audit_sample, sample_manifest, select_random_audit_blocks,
    selected_autoregressive_targets, selected_input_rows, selected_predictions, selected_targets,
)

ref = select_random_audit_blocks(SHARD_ROOT, max_shards=1, samples_per_shard=1, seed=SEED, tickers=TICKERS)[0]
sample = load_audit_sample(SHARD_ROOT, ref)
origin_offset = random.Random(f'{SEED}|{ref.unit_key}|{ref.block_offset}').randrange(sample.block.origin_indices.numel())
manifest = sample_manifest(sample, origin_offset=origin_offset)
display(pl.DataFrame([manifest['reference']]))
display(pl.DataFrame(manifest['context']))
print('selected origin offset:', origin_offset, 'timestamp_us:', manifest['origin_timestamp_us'])
```

## Exact model-ready input rows visible at the selected origin

```python
input_rows = selected_input_rows(sample, origin_offset=origin_offset, tail_rows=TAIL_CONTEXT_ROWS)
for view, rows in input_rows.items():
    print(f'\n{view}: last {len(rows)} causally visible rows')
    if not rows:
        print('unavailable/masked')
        continue
    frame = pl.DataFrame(rows)
    display(frame)
    asof = rows[-1]
    metadata = {'row_index', 'bar_start_us', 'bar_end_us', 'available_at_us', 'is_asof'}
    display(pl.DataFrame({'feature': [name for name in asof if name not in metadata], 'value': [asof[name] for name in asof if name not in metadata]}))
```

## Stored targets and masks

```python
data_config = data_config_for_sample(sample)
horizon_rows = selected_targets(sample, origin_offset, data_config)
ar_rows = selected_autoregressive_targets(sample, origin_offset)
display(pl.DataFrame(horizon_rows))
display(pl.DataFrame(ar_rows))
```

## Optional checkpoint outputs versus targets

```python
if CHECKPOINT is not None and CHECKPOINT.is_file():
    predictions = selected_predictions(sample, origin_offset=origin_offset, checkpoint_path=CHECKPOINT)
    comparison = pl.DataFrame(horizon_rows).join(pl.DataFrame(predictions), on='horizon', how='left')
    display(comparison)
else:
    print('Set CHECKPOINT to a checkpoint_latest.pt to display model predictions beside targets.')
```
